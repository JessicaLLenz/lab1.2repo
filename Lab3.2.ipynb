```python
import polars as pl
```

```python
(reverse := 
 pl.read_csv("./data/ReverseCodingItems.csv")
)
```

```python
rename_dict = {
    col_name: f"{col_name}_{'rev' if coding == 'Yes' else 'reg'}"
    for col_name, coding in zip(reverse['Column Name'], reverse['Needs Reverse Coding?'])
}
rename_dict
```

```python
(health := 
 pl.read_csv("./data/health_survey.csv")
)
health = health.rename(rename_dict)
health
```

```python
health_recode = {
       "Strongly Agree": 5,
       "Somewhat Agree": 4,
        "Neutral": 3,
        "Somewhat Disagree": 2,
        "Strongly Disagree": 1,
        "Neither Agree nor Disagree": 3
        
}

health_numeric = health.with_columns(
        pl.col(pl.Utf8).replace(health_recode) 
)

health_numeric.head()

```

```python
health_recoded = health_numeric.with_columns(
    [
        (6 - pl.col(c)).alias(c.removesuffix('_rev'))
        for c in health_numeric.select(pl.col(r".*_rev$")).columns
    ]
)
```

```python
question_pre = ['F1', 'F2', 'F3', 'F4', 'F5', 'F6']
health_agg = health_recoded.select(
    [
        pl.reduce(
            lambda a, b: a + b,
            exprs = [pl.col(c).fill_null(0) for c in health_recoded.select(pl.col(f"^{p}_.*$")).columns]
        ).alias(f"{p}_total")
        for p in question_pre
    ]
)

print(health_agg.head())
```
